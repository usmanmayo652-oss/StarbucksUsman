```python
# ============================================================
# EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

# 1. IMPORT LIBRARIES
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings

warnings.filterwarnings('ignore')

# ============================================================
# 2. FIND DATASET FILE
# ============================================================

for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))

# Find CSV files automatically
csv_files = []

for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        if filename.endswith('.csv'):
            csv_files.append(os.path.join(dirname, filename))

print("\nCSV Files Found:")
print(csv_files)

# Load the first CSV file
df = pd.read_csv(csv_files[0])

print("\nDataset loaded successfully!")
print("Shape:", df.shape)

# ============================================================
# 3. DISPLAY FIRST 5 ROWS
# ============================================================

print("\nFirst 5 Rows:")
display(df.head())

# ============================================================
# 4. DISPLAY LAST 5 ROWS
# ============================================================

print("\nLast 5 Rows:")
display(df.tail())

# ============================================================
# 5. DATASET SHAPE
# ============================================================

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

# ============================================================
# 6. COLUMN NAMES
# ============================================================

print("\nColumn Names:")
print(df.columns.tolist())

# ============================================================
# 7. DATA TYPES
# ============================================================

print("\nData Types:")
print(df.dtypes)

# ============================================================
# 8. BASIC INFORMATION
# ============================================================

print("\nDataset Information:")
df.info()

# ============================================================
# 9. STATISTICAL SUMMARY
# ============================================================

print("\nStatistical Summary:")
display(df.describe(include='all').T)

# ============================================================
# 10. MISSING VALUES
# ============================================================

print("\nMissing Values:")
missing_values = df.isnull().sum()

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Values': missing_values,
    'Missing Percentage': missing_percentage
})

display(missing_df)

# ============================================================
# 11. DUPLICATE VALUES
# ============================================================

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

# ============================================================
# 12. UNIQUE VALUES
# ============================================================

print("\nNumber of Unique Values in Each Column:")
display(df.nunique().to_frame('Unique Values'))

# ============================================================
# 13. CATEGORICAL COLUMN VALUES
# ============================================================

categorical_columns = df.select_dtypes(include=['object', 'category']).columns

print("\nCategorical Columns:")
print(list(categorical_columns))

for col in categorical_columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts().head(15))

# ============================================================
# 14. NUMERICAL COLUMNS
# ============================================================

numerical_columns = df.select_dtypes(include=np.number).columns

print("\nNumerical Columns:")
print(list(numerical_columns))

# ============================================================
# 15. HISTOGRAMS
# ============================================================

if len(numerical_columns) > 0:

    df[numerical_columns].hist(
        figsize=(15, 12),
        bins=20
    )

    plt.suptitle("Distribution of Numerical Features")
    plt.tight_layout()
    plt.show()

# ============================================================
# 16. BOXPLOTS - OUTLIER DETECTION
# ============================================================

for col in numerical_columns:

    plt.figure(figsize=(8, 4))

    sns.boxplot(x=df[col])

    plt.title(f"Boxplot of {col}")
    plt.xlabel(col)

    plt.show()

# ============================================================
# 17. CORRELATION MATRIX
# ============================================================

if len(numerical_columns) > 1:

    plt.figure(figsize=(12, 8))

    correlation = df[numerical_columns].corr()

    sns.heatmap(
        correlation,
        annot=True,
        cmap='coolwarm',
        fmt='.2f'
    )

    plt.title("Correlation Heatmap")
    plt.show()

# ============================================================
# 18. PAIRPLOT
# ============================================================

if len(numerical_columns) >= 2:

    # Use maximum 5 columns to avoid very large plots
    pair_columns = numerical_columns[:5]

    sns.pairplot(df[pair_columns].dropna())

    plt.show()

# ============================================================
# 19. COUNT PLOTS FOR CATEGORICAL VARIABLES
# ============================================================

for col in categorical_columns:

    # Only plot columns with reasonable number of categories
    if df[col].nunique() <= 15:

        plt.figure(figsize=(10, 5))

        sns.countplot(
            data=df,
            x=col,
            order=df[col].value_counts().index
        )

        plt.title(f"Distribution of {col}")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

# ============================================================
# 20. TOP CATEGORIES
# ============================================================

for col in categorical_columns:

    if df[col].nunique() > 1:

        print(f"\nTop categories in {col}:")

        display(
            df[col]
            .value_counts()
            .head(10)
            .to_frame('Count')
        )

# ============================================================
# 21. SKEWNESS
# ============================================================

if len(numerical_columns) > 0:

    print("\nSkewness of Numerical Features:")

    skewness = df[numerical_columns].skew()

    display(skewness.to_frame('Skewness'))

# ============================================================
# 22. OUTLIER COUNT USING IQR
# ============================================================

print("\nOutlier Count:")

outlier_results = {}

for col in numerical_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_results[col] = len(outliers)

outlier_df = pd.DataFrame(
    list(outlier_results.items()),
    columns=['Column', 'Outlier Count']
)

display(outlier_df)

# ============================================================
# 23. MISSING VALUES VISUALIZATION
# ============================================================

missing_count = df.isnull().sum()

missing_count = missing_count[missing_count > 0]

if len(missing_count) > 0:

    plt.figure(figsize=(10, 5))

    missing_count.sort_values(ascending=False).plot(
        kind='bar'
    )

    plt.title("Missing Values by Column")
    plt.xlabel("Columns")
    plt.ylabel("Missing Values")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()

else:

    print("\nNo missing values found!")

# ============================================================
# 24. DATASET SUMMARY
# ============================================================

print("\n" + "="*50)
print("FINAL EDA SUMMARY")
print("="*50)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Duplicate Rows:", df.duplicated().sum())
print("Total Missing Values:", df.isnull().sum().sum())
print("Numerical Columns:", len(numerical_columns))
print("Categorical Columns:", len(categorical_columns))

print("\nEDA Completed Successfully!")
```

```python

# ============================================================
# DECISION TREE CLASSIFICATION - KAGGLE
# ============================================================

# 1. IMPORT LIBRARIES
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# 2. LOAD DATASET
# ============================================================

csv_files = []

for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        if filename.lower().endswith('.csv'):
            csv_files.append(os.path.join(dirname, filename))

if not csv_files:
    raise FileNotFoundError(
        "No CSV file found in /kaggle/input"
    )

print("Available CSV files:")
for i, path in enumerate(csv_files):
    print(i, path)

# Load the first CSV file
df = pd.read_csv(csv_files[0])

print("\nDataset Loaded Successfully!")
print("Dataset Shape:", df.shape)
display(df.head())

# ============================================================
# 3. CLEAN DATA
# ============================================================

# Remove duplicate rows
df = df.drop_duplicates().copy()

# Replace infinite values with NaN
df = df.replace([np.inf, -np.inf], np.nan)

# ============================================================
# 4. SELECT TARGET COLUMN
# ============================================================

# Change this to your actual prediction column if necessary
TARGET_COLUMN = df.columns[-1]

print("\nTarget Column:", TARGET_COLUMN)
print("Target Values:")
print(df[TARGET_COLUMN].value_counts(dropna=False).head(20))

# Remove rows where the target is missing
df = df.dropna(subset=[TARGET_COLUMN]).copy()

# Remove rows with a target class occurring only once
# This allows a stratified 80/20 split when feasible
target_counts = df[TARGET_COLUMN].value_counts()
rare_classes = target_counts[target_counts < 2].index

if len(rare_classes) > 0:
    print("\nRemoving target classes with fewer than 2 rows:")
    print(list(rare_classes))
    df = df[~df[TARGET_COLUMN].isin(rare_classes)].copy()

if df[TARGET_COLUMN].nunique() < 2:
    raise ValueError(
        "The target needs at least 2 different classes."
    )

X = df.drop(columns=[TARGET_COLUMN]).copy()
y = df[TARGET_COLUMN].copy()

# Remove ID columns if they are unique for every row
id_columns = [
    col for col in X.columns
    if 'id' in col.lower()
    and X[col].nunique() == len(X)
]

if id_columns:
    X = X.drop(columns=id_columns)
    print("\nRemoved ID columns:", id_columns)

# ============================================================
# 5. ENCODE TARGET LABELS
# ============================================================

target_encoder = LabelEncoder()
y = target_encoder.fit_transform(y.astype(str))

print("\nTarget Classes:")
print(list(target_encoder.classes_))

# ============================================================
# 6. IDENTIFY COLUMN TYPES
# ============================================================

numeric_cols = X.select_dtypes(
    include=np.number
).columns

categorical_cols = X.select_dtypes(
    exclude=np.number
).columns

print("\nNumerical Columns:", list(numeric_cols))
print("Categorical Columns:", list(categorical_cols))

# ============================================================
# 7. HANDLE MISSING VALUES
# ============================================================

# Fill numerical missing values using median
for col in numeric_cols:
    X[col] = X[col].fillna(X[col].median())

# Fill categorical missing values using mode
for col in categorical_cols:
    mode_values = X[col].mode()
    fill_value = mode_values.iloc[0] if not mode_values.empty else "Unknown"
    X[col] = X[col].fillna(fill_value).astype(str)

# ============================================================
# 8. ENCODE CATEGORICAL FEATURES
# ============================================================

X = pd.get_dummies(X, drop_first=False, dtype=int)

print("\nFeatures After Encoding:", X.shape[1])
display(X.head())

# ============================================================
# 9. TRAIN-TEST SPLIT (80/20)
# ============================================================

# Stratify only when each class has enough examples
class_counts = pd.Series(y).value_counts()
test_size = max(1, int(np.ceil(0.20 * len(y))))

can_stratify = (
    class_counts.min() >= 2
    and test_size >= len(class_counts)
    and (len(y) - test_size) >= len(class_counts)
)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y if can_stratify else None
)

print("\nTraining Samples:", len(X_train))
print("Testing Samples:", len(X_test))

# ============================================================
# 10. TRAIN DECISION TREE MODEL
# ============================================================

model = DecisionTreeClassifier(
    criterion='gini',
    max_depth=5,
    random_state=42
)

model.fit(X_train, y_train)

print("\nDecision Tree Model Trained Successfully!")

# ============================================================
# 11. MAKE PREDICTIONS
# ============================================================

y_pred = model.predict(X_test)

# ============================================================
# 12. MODEL ACCURACY
# ============================================================

accuracy = accuracy_score(y_test, y_pred)

print("\nDecision Tree Accuracy: {:.2f}%".format(
    accuracy * 100
))

# ============================================================
# 13. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")

print(classification_report(
    y_test,
    y_pred,
    labels=np.arange(len(target_encoder.classes_)),
    target_names=target_encoder.classes_,
    zero_division=0
))

# ============================================================
# 14. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=np.arange(len(target_encoder.classes_))
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=target_encoder.classes_,
    yticklabels=target_encoder.classes_
)

plt.title("Decision Tree Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.tight_layout()
plt.show()

# ============================================================
# 15. VISUALIZE DECISION TREE
# ============================================================

plt.figure(figsize=(22, 12))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=target_encoder.classes_,
    filled=True,
    rounded=True,
    fontsize=8
)

plt.title("Decision Tree Visualization")
plt.show()

# ============================================================
# 16. FEATURE IMPORTANCE
# ============================================================

importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': model.feature_importances_
}).sort_values(
    by='Importance',
    ascending=False
)

print("\nTop 15 Important Features:")
display(importance.head(15))

plt.figure(figsize=(10, 6))

top_features = importance.head(15).sort_values(
    by='Importance'
)

sns.barplot(
    data=top_features,
    x='Importance',
    y='Feature'
)

plt.title("Top 15 Decision Tree Feature Importances")
plt.tight_layout()
plt.show()

# ============================================================
# 17. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 45)
print("DECISION TREE FINAL RESULTS")
print("=" * 45)

print("Dataset Shape:", df.shape)
print("Target Column:", TARGET_COLUMN)
print("Training Samples:", len(X_train))
print("Testing Samples:", len(X_test))
print("Accuracy: {:.2f}%".format(accuracy * 100))
print("Tree Depth:", model.get_depth())
print("Number of Leaves:", model.get_n_leaves())

print("\nDecision Tree Analysis Completed!")
```
